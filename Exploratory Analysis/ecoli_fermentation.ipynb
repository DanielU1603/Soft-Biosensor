```python
#relative difference between spectrum 1 and 2 

mean_absolute_difference = (train_df.loc[0, spectral_columns]
                            - train_df.loc[1, spectral_columns]).abs().mean()

mean_intensity = (
    train_df.loc[0, spectral_columns].mean() 
    +  train_df.loc[1, spectral_columns].mean())/2

relative_difference = (mean_absolute_difference/mean_intensity)*100

print(relative_difference)
```

```python
#checking interval between raman shifts

df_numeric_columns = df.columns[:-2]
raman_shift = df_numeric_columns.astype("float32")



interval_size = pd.Series(raman_shift).diff().dropna()
interval_size_pandas_series = pd.Series(raman_shift).diff()
min_value = interval_size.min()
max_value = interval_size.max()

interval_size.describe()

#this step was implemented to check if interval size decreasing was gradual following a pattern
change_point = interval_size[interval_size < 2.0]
print("First 20 Change Points in original interval size:", interval_size.iloc[310:335])


```

```python
#plotting interval sizes along raman shift axis 

plt.figure(figsize=(12, 6))

plt.plot(interval_size)
plt.title("Interval Size Difference vs Interval Index")
plt.xlabel("Interval Index")
plt.ylabel("Interval Size Difference")
plt.show()
```

```python
#relationship between variables 

raman_shift = df.loc[:, "390.63"]
glucose_conc = df.loc[:, "Glucose"]
acetate_conc = df.loc[:, "Acetate"]


#configuration of dispersion plot 

plt.figure(figsize=(6,4))

plt.scatter(raman_shift, glucose_conc, marker="o", color="red", label="Glucose Concentration")
plt.scatter(raman_shift, acetate_conc, marker="s", color="green", label="Acetate Concentration")
plt.title("Relation between raman shift and metabolites")
plt.xlabel("raman_shift")
plt.ylabel("Metabolite Concentration")
plt.legend()
plt.show()

#we should further explore what's happening with glucose concentration
#because we can distinguish between two regions for red points.
#the first ones are completely at the bottom of the graph, whereas the others spread 
#around higher values of concentrations
#It's valuable to wonder if this clustering is caused by the different stages of microorganism's growing
```

```python
#visualizing raman_spectra

#first spectra
raman_intensity_first_spectra = df.iloc[0][:-2]
intensity_shape = raman_intensity_first_spectra.shape
raman_signals = df.columns[:-2]
signal_shape = raman_signals.shape

#spectrum number 100
raman_intensity_spectrum_100 = df.iloc[100][:-2]
print(f"Signal shape: {signal_shape} and Intensity_shape: {intensity_shape}")

#plot configuration

plt.figure(figsize=(16, 12))
plt.plot(raman_signals, raman_intensity_first_spectra, marker="o", color="gold", linestyle="-", linewidth=0.5, label="Raman Spectrum Number 1")
plt.plot(raman_signals, raman_intensity_spectrum_100, marker="s", color="lightcoral", linestyle="-", linewidth=0.5, label="Raman Spectrum Number 100" )

plt.title("First Raman Spectra")
plt.xlabel("Raman Signal Value")
plt.ylabel("Raman Signal Intensity")

plt.legend() 
plt.show()
```

```python
#finding glucose signal in raman spectrum using column filters

glucose_signals = []
acetate_signals = []

df_columns = df.columns[:-2].astype("float64")

for signal in df_columns: 
    if signal > 900 and signal < 950:
        glucose_signals.append(signal)

    if signal >= 1060 and signal <= 1130:
        acetate_signals.append(signal)

print(f"These are the glucose signals: {glucose_signals}")
print(f"These are the acetate signals: {acetate_signals}")


glucose_id_signal = df["920.07"]
acetate_id_signal_1 = df["1060.19"]
acetate_id_signal_2 = df["1119.87"]

print(glucose_id_signal)
```

```python
#multiple plots 

gs = gsp.GridSpec(2, 2, height_ratios=(1, 1), width_ratios=(1, 1))

#first plot 
fig = plt.figure(figsize=(10, 8))

ax1 = fig.add_subplot(gs[1, 0])
ax1.plot(raman_signals, raman_intensity_first_spectra, marker="o", color="lightcoral", linestyle="-", linewidth=0.5, label="First Raman Spectrum")
ax1.set_title("First Raman Spectrum")
ax1.set_xlabel("Raman Shift Value")
ax1.set_ylabel("Raman Signal Intensity")

ax2 = fig.add_subplot(gs[1, 1])
ax2.plot(raman_signals, raman_intensity_spectrum_100, marker="s", color="gold", linestyle="--", linewidth=0.5, label ="Raman Spectrum number 100")
ax2.set_title("Raman Spectrum 100")
ax2.set_xlabel("Raman Shift Value")
ax2.set_ylabel("Raman Signal Intensity")

plt.tight_layout()
plt.show()
```
